```python
import pandas as pd

students = pd.read_csv("student_marks_messy.csv")
titanic = pd.read_csv("titanic.csv")

display(students.head())
display(titanic.head())
```

1a

```python
for name, df in [("Students", students), ("Titanic", titanic)]:
    print(f"\n--- {name} ---")
    print("Rows and columns:", df.shape)

    print("\nColumn information:")
    df.info()

    print("\nSummary statistics:")
    display(df.describe(include="all"))

    print("\nMissing values:")
    display(df.isnull().sum())
```

1b

```python
# Passengers older than 60
display(titanic.loc[titanic["Age"] > 60])

# First 5 rows and first 4 columns
display(titanic.iloc[:5, :4])
```

1c

```python
# Standardize gender values
students["gender"] = (
    students["gender"]
    .str.strip()
    .str.lower()
    .replace({
        "m": "Male",
        "male": "Male",
        "f": "Female",
        "female": "Female"
    })
)

# Convert English marks to numbers
# Invalid text becomes NaN (a missing value)
students["english_marks"] = pd.to_numeric(
    students["english_marks"], errors="coerce"
)

# Remove duplicate rows
before = len(students)
students = students.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed:", before - len(students))
display(students)
```

1d

```python
# Fill missing marks with each subject's median
mark_columns = ["math_marks", "science_marks", "english_marks"]

for column in mark_columns:
    students[column] = students[column].fillna(
        students[column].median()
    )

# Fill missing ages with the median age of each passenger class
titanic["Age"] = titanic["Age"].fillna(
    titanic.groupby("Pclass")["Age"].transform("median")
)

# Check that the missing values have been filled
print("Missing marks:")
print(students[mark_columns].isnull().sum())

print("\nMissing ages:", titanic["Age"].isnull().sum())
```

1e

```python
class_summary = titanic.groupby("Pclass").agg(
    mean_fare=("Fare", "mean"),
    mean_age=("Age", "mean"),
    survival_rate=("Survived", "mean")
)

display(class_summary.round(2))
```

1f

```python
# Create a lookup table
ports = pd.DataFrame({
    "Embarked": ["C", "Q", "S"],
    "Port_Name": ["Cherbourg", "Queenstown", "Southampton"]
})

# Add port names to the passenger records
titanic_with_ports = titanic.merge(
    ports,
    on="Embarked",
    how="left",
    validate="many_to_one"
)

display(titanic_with_ports[["Name", "Embarked", "Port_Name"]])
```

```python
students.to_csv("student_marks_cleaned.csv", index=False)
titanic_with_ports.to_csv("titanic_cleaned.csv", index=False)

print("Both cleaned files saved!")
```
